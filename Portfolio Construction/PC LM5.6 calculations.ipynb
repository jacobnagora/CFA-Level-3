```python
# Portfolio Construction - Learning Module 5.6 - Portfolio Management for Institutional Investors - Balance Sheet Management and Investment Considerations

def pct_chg_val_equity (pct_e, pct_l,lev_r ):
    # Calculate expected percentage change in the MV of equity
    return (pct_e * lev_r) - (pct_l*(lev_r-1))

# test - should return -.285
pct_chg_val_equity(pct_e=0,pct_l=0.015,lev_r=1/0.05)
```

```python
def pct_chg_equity_dur(md_a,md_l,lev_r,chg_yl_ya):
    """ 
    Calculate expected ModDur of equity capital, where:
    md_a = Modified Duration of Assets
    md_l = Modified Duration of Liabilities
    lev_r = Leverate ratio
    chg_yl_ya = estimated change in yield of liabilities relative to a unit change in yield of assets
    """
    return (md_a*lev_r) - (md_l*(lev_r-1)*chg_yl_ya)

# test - should return 14.2
pct_chg_equity_dur(lev_r=1/.125,md_a=3,md_l=2,chg_yl_ya=70/10000/0.01)
```

```python
def pct_chg_std_dev_equity (std_a, std_l,lev_r, corr_pct_chg):
    # Calculate expected percentage change in standard deviation of the MV of equity
    return (lev_r**2 * std_a**2) + (((lev_r-1)**2)*std_l**2) - (2*lev_r*(lev_r-1)*std_a*std_l*corr_pct_chg)

# test - should return Variance of 0.298
test = pct_chg_std_dev_equity(std_a = 0.07,
std_l = 0.04,
lev_r= 1/0.125,
corr_pct_chg= 0.3)
print(test)

# Asset volatility decreases to 5% and Equity-to-Assets increases to 20%. Calculate Variance and StdDev.
import math

var = pct_chg_std_dev_equity(std_a = 0.05,
std_l = 0.04,
lev_r= 1/0.2,
corr_pct_chg= 0.3)   
var, math.sqrt(var)
```
